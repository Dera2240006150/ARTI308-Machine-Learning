```python
# --- Step 1: Import Libraries ---
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style for plots
sns.set_style('whitegrid')
%matplotlib inline

# --- Step 2: Load the Data ---
# Make sure Housing.csv is uploaded to your Colab files
df = pd.read_csv('Housing.csv')

# --- Step 3: Understand the Data Structure ---
print("--- First 5 Rows of Data ---")
display(df.head())

print("\n--- Data Info (Types & Nulls) ---")
print(df.info())

print("\n--- Statistical Summary (Numerical) ---")
display(df.describe())

print("\n--- Dataset Shape ---")
print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")

# --- Step 4: Data Cleaning Checks ---
print("\n--- Checking for Missing Values ---")
print(df.isnull().sum())

print("\n--- Checking for Duplicates ---")
duplicates = df.duplicated().sum()
print(f"Duplicate rows: {duplicates}")
# Optional: df.drop_duplicates(inplace=True) if duplicates > 0

# --- Step 5: Univariate Analysis (Distribution of single variables) ---

# 5.1 Distribution of the Target Variable (Price)
plt.figure(figsize=(10, 6))
sns.histplot(df['price'], kde=True, bins=30, color='blue')
plt.title('Distribution of House Prices')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.show()

# 5.2 Distribution of Area
plt.figure(figsize=(10, 6))
sns.histplot(df['area'], kde=True, bins=30, color='green')
plt.title('Distribution of Area (sq ft)')
plt.show()

# 5.3 Categorical Counts (e.g., Furnishing Status)
plt.figure(figsize=(8, 5))
sns.countplot(x='furnishingstatus', data=df, palette='viridis')
plt.title('Count of Furnishing Status')
plt.show()

# --- Step 6: Bivariate Analysis (Relationships between variables) ---

# 6.1 Numerical vs Numerical: Price vs Area
plt.figure(figsize=(10, 6))
sns.scatterplot(x='area', y='price', data=df, alpha=0.6)
plt.title('Price vs. Area')
plt.show()

# 6.2 Categorical vs Numerical: Price vs Bedrooms (Boxplot for Outliers)
plt.figure(figsize=(10, 6))
sns.boxplot(x='bedrooms', y='price', data=df, palette='coolwarm')
plt.title('House Price Distribution by Number of Bedrooms')
plt.show()

# 6.3 Categorical vs Numerical: Price vs Air Conditioning
plt.figure(figsize=(8, 5))
sns.boxplot(x='airconditioning', y='price', data=df, palette='Set2')
plt.title('Effect of Air Conditioning on Price')
plt.show()

# --- Step 7: Correlation Analysis (Heatmap) ---

# To create a correlation heatmap, we need to convert text categories (yes/no) to numbers
# Creating a copy for correlation analysis
df_corr = df.copy()

# Mapping binary variables to 1 and 0
binary_cols = ['mainroad', 'guestroom', 'basement', 'hotwaterheating', 'airconditioning', 'prefarea']
def binary_map(x):
    return x.map({'yes': 1, 'no': 0})

df_corr[binary_cols] = df_corr[binary_cols].apply(binary_map)

# Mapping furnishingstatus to numbers just for correlation check
df_corr['furnishingstatus'] = df_corr['furnishingstatus'].map({'furnished': 2, 'semi-furnished': 1, 'unfurnished': 0})

# Plotting the Heatmap
plt.figure(figsize=(12, 10))
sns.heatmap(df_corr.corr(), annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Correlation Matrix of Housing Features')
plt.show()
```
