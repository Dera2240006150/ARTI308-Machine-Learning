```python
# --- Imports ---
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set the visual theme for Seaborn
sns.set_theme(style="whitegrid")

# --- Load the Data ---
df = pd.read_csv('Housing.csv')

# --- A. Distribution Plots ---

# 1. Distplot (Distribution of a single variable)
sns.displot(df['price'], kde=True, color='darkblue', bins=30)
plt.title('Distribution of Prices')
plt.show()

# 2. Jointplot (Compare two distributions)
sns.jointplot(x='area', y='price', data=df, kind='scatter', color='teal', alpha=0.6)
plt.show()

# 3. Pairplot (Quick visual of relationships across all numeric columns)
sns.pairplot(df[['price', 'area', 'bedrooms', 'bathrooms']])
plt.show()

# 4. Rugplot and KDE Plot
plt.figure(figsize=(8, 5))
sns.kdeplot(df['area'], fill=True, color='red')
sns.rugplot(df['area'], color='black')
plt.title('KDE and Rugplot for Area')
plt.show()


# --- B. Categorical Plots ---

# 5. Barplot (Shows the mean of a quantitative variable per category)
plt.figure(figsize=(8, 5))
sns.barplot(x='furnishingstatus', y='price', data=df, palette='coolwarm')
plt.title('Average Price by Furnishing Status')
plt.show()

# 6. Countplot (Counts the occurrences of a categorical variable)
plt.figure(figsize=(8, 5))
sns.countplot(x='airconditioning', data=df, palette='Set2')
plt.title('Count of Houses with/without Air Conditioning')
plt.show()

# 7. Boxplot and Violinplot
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.boxplot(x='mainroad', y='price', data=df, palette='pastel')
plt.title('Price Boxplot based on Main Road')

plt.subplot(1, 2, 2)
sns.violinplot(x='mainroad', y='price', data=df, palette='pastel')
plt.title('Price Violinplot based on Main Road')
plt.tight_layout()
plt.show()

# 8. Stripplot and Swarmplot
plt.figure(figsize=(10, 5))
sns.stripplot(x='bedrooms', y='price', data=df, jitter=True, alpha=0.6, color='black')
sns.violinplot(x='bedrooms', y='price', data=df, inner=None, color='lightgray')
plt.title('Strip Plot & Violin Plot of Price per Bedroom Count')
plt.show()


# --- C. Matrix Plots & Regression ---

# 9. Heatmap (Correlation Matrix)
corr_matrix = df.select_dtypes(include=[np.number]).corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', linewidths=1)
plt.title('Correlation Heatmap')
plt.show()

# 10. lmplot (Linear Regression Plot)
sns.lmplot(x='area', y='price', data=df, hue='airconditioning', palette='Set1', aspect=1.5)
plt.title('Linear Regression: Area vs Price (Colored by Air Conditioning)')
plt.show()
```
