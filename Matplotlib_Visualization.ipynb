```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# 1. Load the Dataset
df = pd.read_csv('Housing.csv')

# 2. Basic Line Plot
df_sorted = df.sort_values(by='area')
plt.figure(figsize=(10, 6))
plt.plot(df_sorted['area'], df_sorted['price'], color='blue', alpha=0.5)
plt.title('House Price vs Area (Line Plot)')
plt.xlabel('Area (sq ft)')
plt.ylabel('Price')
plt.show()

# 3. Scatter Plot
plt.figure(figsize=(8, 5))
plt.scatter(df['area'], df['price'], color='red', marker='o', alpha=0.6)
plt.title('Scatter Plot of Price vs Area')
plt.xlabel('Area (sq ft)')
plt.ylabel('Price')
plt.grid(True)
plt.show()

# 4. Histogram
plt.figure(figsize=(8, 5))
plt.hist(df['price'], bins=30, color='green', edgecolor='black')
plt.title('Distribution of House Prices')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.show()

# 5. Boxplot
plt.figure(figsize=(8, 5))
plt.boxplot(df['price'], vert=False, patch_artist=True)
plt.title('Boxplot of House Prices')
plt.xlabel('Price')
plt.show()

# 6. Subplots (Multiple plots in one figure)
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(12, 5))

# First subplot
axes[0].scatter(df['bedrooms'], df['price'], color='purple', alpha=0.5)
axes[0].set_title('Price vs Bedrooms')
axes[0].set_xlabel('Bedrooms')
axes[0].set_ylabel('Price')

# Second subplot
axes[1].hist(df['area'], bins=20, color='orange', edgecolor='black')
axes[1].set_title('Distribution of Area')
axes[1].set_xlabel('Area (sq ft)')

plt.tight_layout()
plt.show()
```
